```python
import pandas as pd 
```

```python
d = {
    'sales': [125000, 235000, 180000, 325000, 95000, 450000, 275000, 150000, 390000, 210000, 520000, 175000],
    
    'city': ['Tampa', 'Orlando', 'Miami', 'Jacksonville', 'Tampa', 'Miami', 
             'Orlando', 'Jacksonville', 'Miami', 'Tampa', 'Orlando', 'Jacksonville'],
    
    'size': ['Small', 'Medium', 'Large', 'Large', 'Small', 'Medium', 
             'Large', 'Small', 'Medium', 'Medium', 'Large', 'Small']
}
```

```python
df = pd.DataFrame(data = d)
```

```python
df.head()
```

```python
df["city"].unique()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(handle_unknown = 'ignore', sparse_output = False).set_output(transform = 'pandas')
```

```python
ohetransform = ohe.fit_transform(df[["city"]])
```

```python
ohetransform
```

```python
 df.head()
```

```python
ohetransform.head()
```

```python
df = pd.concat([df, ohetransform], axis = 1).drop(columns = ["city"])
```
