```python
import pandas as pd
```

```python
df = pd.read_csv("medical_fixed.csv", encoding = "latin-1")
```

```python
df.head()
```

```python
df = df.drop(columns = ["sex", "smoker", "region", "charges"])
```

```python
df.info()
```

```python
df.describe().round(3)
```

```python
X1 = df.iloc[:,:] # standard scaling
```

```python
X2 = df.iloc[:,:] # min/max scaling
```

```python
# STANDARDIZATION
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaleStandard = StandardScaler()
```

```python
X1 = scaleStandard.fit_transform(X1)
```

```python
X1 = pd.DataFrame(X1, columns = ["age", "bmi", "children"])
```

```python
X1.head()
```

```python
X1.describe().round(3)
```

```python
# NORMALIZATION
```

```python
from sklearn.preprocessing import MinMaxScaler                                                            
```

```python
scaleMinMax = MinMaxScaler(feature_range = (0,1))
```

```python
X2 = scaleMinMax.fit_transform(X2)
```

```python
X2 = pd.DataFrame(X2, columns = ["age", "bmi", "children"])
```

```python
X2.head()
```

```python
X2.describe().round(3)
```
